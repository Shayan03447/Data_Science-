```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
```

```python
data=load_iris()
```

```python
data.keys()
```

```python
df=pd.DataFrame(data=data['data'],columns=data['feature_names'])
```

```python
df.head()
```

```python
df['Target']=data['target']
```

```python
df.head()
```

```python
data['target_names']
```

```python
df["Flower_name"]=df['Target'].apply(lambda x: data['target_names'][x])
```

```python
df.head()
```

```python
df[df['Target']==1]
```

```python
X=df.iloc[:,0:-2]
X
```

```python
y=df.iloc[:,-2]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=101)
```

```python
len(X_train)
```

```python
len(y_test)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
bagging_model=BaggingClassifier(estimator=DecisionTreeClassifier(),n_estimators=5)
bagging_model.fit(X_train_scaled,y_train)
```

```python
y_pred=bagging_model.predict(X_test_scaled)
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.ensemble import AdaBoostClassifier
AdaBoost=AdaBoostClassifier(n_estimators=10)
```

```python
AdaBoost.fit(X_train_scaled,y_train)
```

```python
y_predict=AdaBoost.predict(X_test_scaled)
```

```python
accuracy_score(y_test,y_predict)
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
```

```python
#define base model
base_models=[
    ('tree', DecisionTreeClassifier()),
    ('svm',SVC(probability=True))
]
stacking_model=StackingClassifier(estimators=base_models,final_estimator=LogisticRegression())
stacking_model.fit(X_train_scaled,y_train)
```

```python
y_pre=stacking_model.predict(X_test)
```

```python
accuracy_score(y_test,y_pre)
```
