Ensamble_learning:combination of multiple models to make more accurate robust and stable pridiction

```python
X=df.iloc[:,0:-1]
X
```

```python
y=df.iloc[:,-1]
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_scaled_train,X_scaled_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=101)
```

```python
#Bagging (Bootstrap Aggregating)
#Famous random_forest
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=100)
```

```python
rf.fit(X_scaled_train,y_train)
```

```python
y_prediction=rf.predict(X_scaled_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_prediction)
```

```python
#Boosting(Sequantial training of models)
#Each model focus on the errors of pervious model
#Adaboost
#Gradient Boosting
#XGBoost
```

```python
#AdaBoost
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
model=AdaBoostClassifier(DecisionTreeClassifier(max_depth=1),n_estimators=50)
model.fit(X_scaled_train,y_train)
```

```python
y_prediction_1=model.predict(X_scaled_test)
```

```python
accuracy_score(y_test,y_prediction_1)
```

```python
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
#Base model
model1=LogisticRegression()
model2=SVC(probability=True)#For softvoting probablity should be must 
model3=DecisionTreeClassifier()
```

```python
#create a voting classifier
voting=VotingClassifier(
    estimators=[('lr',model1),('svm',model2),('dt',model3)],
    voting='soft')
```

```python
voting.fit(X_scaled_train,y_train)
```

```python
y_prediction2=voting.predict(X_scaled_test)
```

```python
accuracy_score(y_test,y_prediction2)
```

# Stacking Classifier

```python
from sklearn.ensemble import StackingClassifier
```

```python
#Base model that are trained on first level
#pridiction of base model and the output goes to the nest level
base_model=[
    ('dt',DecisionTreeClassifier()),
    ('svc',SVC(probability=True))
]
```

```python
mata_model=LogisticRegression()
#mata_model use the prevoius output as an input
```

```python
stack=StackingClassifier(estimators=base_model,final_estimator=mata_model)
```

```python
stack.fit(X_scaled_train,y_train)
```

```python
y_prediction3=stack.predict(X_scaled_test)
```

```python
accuracy_score(y_test,y_prediction3)
```
