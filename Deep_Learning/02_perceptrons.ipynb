```python
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y=make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                        n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,
                        class_sep=10)
```

np(numpy).insert(insert the value)(X(input data rows and columns),0(index means insert at the start of every row),1 (inserted value),axis=1 means insert columns wise)

```python
plt.scatter(X[:,0], X[:,1], c=y, cmap="winter", s=100)
```

```python
def step(x):
    return 1 if x >= 0 else 0

def perceptron(X,y):
    X=np.insert(X,0,1,axis=1)
    weights=np.ones(X.shape[1])
    lr=0.1

    for i in range(1000):
        j=np.random.randint(0,100)
        y_hat=step(np.dot(X[j],weights))
        weights=weights+lr*(y[j]-y_hat*X[j])
    return weights[0],weights[1:]
```

```python
intercept_,coef_=perceptron(X,y)
```

```python
print(intercept_)
print(coef_)
```

```python
m=-(coef_[0]/coef_[1])
b=-(intercept_/coef_[1])
```

```python
x_input=np.linspace(-3,3,100)
y_input=m*x_input+b
```
