```python
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y=make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                        n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,
                        class_sep=10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=30)
```

```python
def perceptron(X, y):
    w1 = w2 = b = 1
    lr = 0.1
    for j in range(1000):
        for i in range(X.shape[0]):
            # check condition
            z = w1 * X[i][0] + w2 * X[i][1] + b
            if z * (1 if y[i] == 1 else -1) < 0:
                # Update weights and bias (example update rule)
                w1 = w1 + lr * (1 if y[i] == 1 else -1) * X[i][0]
                w2 = w2 + lr * (1 if y[i] == 1 else -1) * X[i][1]
                b = b + lr * (1 if y[i] == 1 else -1)
                

                #w1=w1+lr*y[i]*x[i][0]
                #w2=w2+lr*y[i]*x[i][1]
                #b=b+lr*y[i]
        return w1,w2,b
```

```python
w1,w2,b=perceptron(X,y)
```
